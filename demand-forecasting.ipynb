```python
split_idx = int(len(df) * 0.8)

X = df[features]
y = df[target]

X_train = X.iloc[:split_idx]
X_test = X.iloc[split_idx:]

y_train = y.iloc[:split_idx]
y_test = y.iloc[split_idx:]

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))

print("Training dates:",
      df["dteday"].iloc[0],
      "to",
      df["dteday"].iloc[split_idx - 1])

print("Testing dates:",
      df["dteday"].iloc[split_idx],
      "to",
      df["dteday"].iloc[-1])
```

```python

baseline = DummyRegressor(strategy="mean")
```

```python
baseline.fit(X_train, y_train)
```

```python
baseline_predictions = baseline.predict(X_test)

print("Baseline prediction:",
      round(baseline_predictions[0], 2))
```

```python

def evaluate_model(actual, predicted):
    mae = mean_absolute_error(actual, predicted)

    rmse = np.sqrt(
        mean_squared_error(actual, predicted)
    )

    wape = (
        np.abs(actual - predicted).sum()
        / np.abs(actual).sum()
    ) * 100

    return {
        "MAE": round(mae, 2),
        "RMSE": round(rmse, 2),
        "WAPE (%)": round(wape, 2)
    }

baseline_metrics = evaluate_model(
    y_test.to_numpy(),
    baseline_predictions
)

print("Baseline metrics:", baseline_metrics)
```

```python

def segment_metrics(group):
    actual = group["actual"]
    predicted = group["predicted"]

    return pd.Series({
        "observations": len(group),
        "actual_avg": actual.mean(),
        "predicted_avg": predicted.mean(),
        "MAE": mean_absolute_error(actual, predicted),
        "RMSE": np.sqrt(
            mean_squared_error(actual, predicted)
        ),
        "WAPE_pct": (
            group["absolute_error"].sum()
            / actual.abs().sum() * 100
            if actual.abs().sum() != 0 else np.nan
        )
    })

workingday_comparison = test_results.groupby(
    "workingday"
).apply(segment_metrics, include_groups=False)

workingday_comparison.index = [
    "Non-working day" if x == 0 else "Working day"
    for x in workingday_comparison.index
]

display(workingday_comparison.round(2))
```

```python

import numpy as np
import pandas as pd
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Create a chronological validation split from training data
cut = int(len(X_train) * 0.80)

X_subtrain = X_train.iloc[:cut]
y_subtrain = y_train.iloc[:cut]
X_val = X_train.iloc[cut:]
y_val = y_train.iloc[cut:]

# Train alternative model
extra_model = ExtraTreesRegressor(
    n_estimators=200,
    min_samples_leaf=2,
    max_features=1.0,
    random_state=42,
    n_jobs=-1
)

extra_model.fit(X_subtrain, y_subtrain)
val_pred = extra_model.predict(X_val)

# Evaluation function
def evaluate_model(y_true, y_pred, label):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    wape = np.abs(y_true - y_pred).sum() / np.abs(y_true).sum() * 100
    bias = np.mean(y_pred - y_true)

    high = y_true >= 700

    print(f"\n{label}")
    print(f"MAE: {mae:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"WAPE: {wape:.2f}%")
    print(f"Bias: {bias:.2f}")

    if high.sum() > 0:
        print(f"High-demand observations: {high.sum()}")
        print(f"High-demand MAE: "
              f"{mean_absolute_error(y_true[high], y_pred[high]):.2f}")
        print(f"High-demand bias: "
              f"{np.mean(y_pred[high] - y_true[high]):.2f}")
        print(f"High-demand underprediction rate: "
              f"{(y_pred[high] < y_true[high]).mean() * 100:.2f}%")

evaluate_model(y_val, val_pred, "Extra Trees — Validation")
```

```python

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd

actual = np.asarray(y_val)
high_mask = actual >= 700
results = []

configs = [
    {"max_depth": None, "min_samples_leaf": 1},
    {"max_depth": 20, "min_samples_leaf": 2},
    {"max_depth": 16, "min_samples_leaf": 2},
    {"max_depth": 12, "min_samples_leaf": 2},
    {"max_depth": 16, "min_samples_leaf": 4},
    {"max_depth": 12, "min_samples_leaf": 4},
]

for config in configs:
    model = RandomForestRegressor(
        n_estimators=200,
        max_depth=config["max_depth"],
        min_samples_leaf=config["min_samples_leaf"],
        random_state=42,
        n_jobs=-1
    )
    model.fit(X_subtrain, y_subtrain)
    pred = model.predict(X_val)

    results.append({
        **config,
        "MAE": mean_absolute_error(actual, pred),
        "RMSE": np.sqrt(mean_squared_error(actual, pred)),
        "WAPE (%)": (
            np.abs(actual - pred).sum()
            / np.abs(actual).sum() * 100
        ),
        "Bias": np.mean(pred - actual),
        "High-demand MAE": mean_absolute_error(
            actual[high_mask], pred[high_mask]
        ),
        "High-demand underprediction (%)": (
            np.mean(pred[high_mask] < actual[high_mask]) * 100
        )
    })

tuning_results = pd.DataFrame(results).sort_values(
    ["High-demand MAE", "MAE"]
)
display(tuning_results.round(2))
```

```python

import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

# Refit the original configuration on the subtraining period
capacity_rf = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)
capacity_rf.fit(X_subtrain, y_subtrain)

actual = np.asarray(y_val)
base_pred = capacity_rf.predict(X_val)
high_mask = actual >= 700

rows = []

for buffer in [0, 10, 20, 30, 40, 50]:
    # Operational forecast with a fixed capacity buffer
    forecast = np.maximum(0, base_pred + buffer)
    error = forecast - actual

    rows.append({
        "Buffer (bikes)": buffer,
        "MAE": np.mean(np.abs(error)),
        "RMSE": np.sqrt(np.mean(error ** 2)),
        "Bias": np.mean(error),
        "Underprediction rate (%)": np.mean(error < 0) * 100,
        "Overprediction rate (%)": np.mean(error > 0) * 100,
        "High-demand MAE": np.mean(
            np.abs(error[high_mask])
        ),
        "High-demand underprediction (%)": np.mean(
            forecast[high_mask] < actual[high_mask]
        ) * 100
    })

display(pd.DataFrame(rows).round(2))
```

```python

import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

# Business assumptions
SHORTAGE_COST = 110
SURPLUS_COST = 20
BUFFERS = [0, 10, 20, 30, 40, 50]

# Fit only on the subtraining period
cost_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)
cost_model.fit(X_subtrain, y_subtrain)

# Validation predictions only
actual = np.asarray(y_val, dtype=float)
base_forecast = cost_model.predict(X_val)

# Store row-level costs for the requested +20 scenario
buffer = 20
planned_capacity = np.ceil(
    np.maximum(0, base_forecast + buffer)
)

shortage = np.maximum(0, actual - planned_capacity)
surplus = np.maximum(0, planned_capacity - actual)

row_cost = (
    SHORTAGE_COST * shortage
    + SURPLUS_COST * surplus
)

validation_costs_20 = pd.DataFrame({
    "actual_demand": actual,
    "base_forecast": base_forecast,
    "planned_capacity": planned_capacity,
    "buffer": buffer,
    "shortage_bikes": shortage,
    "surplus_bikes": surplus,
    "shortage_cost": SHORTAGE_COST * shortage,
    "surplus_cost": SURPLUS_COST * surplus,
    "total_cost": row_cost
})

print("Validation rows:", len(validation_costs_20))
print("Total estimated cost:", round(row_cost.sum(), 2))
print("Average cost per bike-hour:", round(row_cost.mean(), 2))
print("Rows with shortages:", int((shortage > 0).sum()))
print("Rows with surplus:", int((surplus > 0).sum()))
print("Total shortage units:", round(shortage.sum(), 2))
print("Total surplus units:", round(surplus.sum(), 2))

display(validation_costs_20.head(10))
```

```python

summary = []

for buffer in BUFFERS:
    capacity = np.ceil(
        np.maximum(0, base_forecast + buffer)
    )

    shortage = np.maximum(0, actual - capacity)
    surplus = np.maximum(0, capacity - actual)

    shortage_cost = SHORTAGE_COST * shortage
    surplus_cost = SURPLUS_COST * surplus
    total_cost = shortage_cost + surplus_cost

    summary.append({
        "Buffer": buffer,
        "Total cost": total_cost.sum(),
        "Average cost per hour": total_cost.mean(),
        "Shortage cost": shortage_cost.sum(),
        "Surplus cost": surplus_cost.sum(),
        "Shortage hours (%)": (shortage > 0).mean() * 100,
        "Surplus hours (%)": (surplus > 0).mean() * 100,
        "Total shortage bikes": shortage.sum(),
        "Total surplus bikes": surplus.sum()
    })

cost_comparison = (
    pd.DataFrame(summary)
    .sort_values("Average cost per hour")
    .reset_index(drop=True)
)

display(cost_comparison.round(2))
```

```python

import numpy as np
import pandas as pd

shortage_costs = [50, 80, 110, 150, 200]
surplus_costs = [10, 20, 40, 60]
buffers = [0, 10, 20, 30, 40, 50]

rows = []

for sc in shortage_costs:
    for pc in surplus_costs:
        for buffer in buffers:
            capacity = np.ceil(
                np.maximum(0, base_forecast + buffer)
            )

            shortage = np.maximum(0, actual - capacity)
            surplus = np.maximum(0, capacity - actual)

            total_cost = (
                sc * shortage.sum()
                + pc * surplus.sum()
            )

            rows.append({
                "Shortage cost": sc,
                "Surplus cost": pc,
                "Buffer": buffer,
                "Average cost/hour": total_cost / len(actual)
            })

sensitivity = pd.DataFrame(rows)

# Find the lowest-cost tested buffer for each cost assumption
best_buffers = (
    sensitivity.sort_values("Average cost/hour")
    .groupby(["Shortage cost", "Surplus cost"])
    .first()
    .reset_index()
)

display(best_buffers.round(2))
```
