EDA

Chart-1 Average rentals by hour

```python
import matplotlib.pyplot as plt
hourly_demand=df.groupby("hr")["cnt"].mean()

plt.figure(figsize=(12,5))
plt.plot(
    hourly_demand.index,
    hourly_demand.values,
    marker='o'
)

plt.title("Average Bike rentals by hour")
plt.xlabel("hour of the day")
plt.ylabel("Average Rentals")
plt.xticks(range(24))
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
print("Peak demand hour:", hourly_demand.idxmax())
print("Average rentals at peak:", round(hourly_demand.max(), 2))
print("Lowest demand hour:", hourly_demand.idxmin())
print("Average rentals at low:", round(hourly_demand.min(), 2))
```

Chart 2 Average rentals by weekday

```python
weekday_demand=df.groupby("weekday")["cnt"].mean()
weekday_names={
    0: "Sunday",
    1: "Monday",
    2: "Tuesday",
    3: "Wednesday",
    4: "Thursday",
    5: "Friday",
    6: "Saturday"
}
weekday_demand.index=[
    weekday_names[d] for d in weekday_demand.index
]
plt.figure(figsize=(10, 5))
weekday_demand.plot(kind="bar")

plt.title("Average bike rentals by weekday")
plt.xlabel("Day of the week")
plt.ylabel("Average rentals")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
print("Highest-demand weekday:")
print(weekday_demand.idxmax(),
    round(weekday_demand.max(),2))
print("lowest-demand weekday:")
print(weekday_demand.idxmin(),
    round(weekday_demand.min(),2))
```

```python
workingday_demand=df.groupby("workingday")["cnt"].mean()

print("Non-working day average:", round(workingday_demand.loc[0],2))
print("working day average:", round(workingday_demand.loc[1],2))
```

Chart 3: Weather and temperature

```python
weather_demand = df.groupby("weathersit")["cnt"].mean()

weather_names={
    1:"clear/partly cloud",
    2:"mist/cloudy",
    3:"light rain/snow",
    4:"heavy rain/snow"
}
weather_demand.index=[
    weather_names[w] for w in weather_demand.index
]

plt.figure(figsize = (10,5))
weather_demand.plot(kind="bar")

plt.title("Average bike rentals by weather")
plt.xlabel("weather conditions")
plt.ylabel("Average rentals")
plt.xticks(rotation=25, ha="right")
plt.grid(axis="y",alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
print(weather_demand.round(2))
```

```python
plt.figure(figsize = (9,5))
plt.scatter(
    df["temp"],
    df["cnt"],
    alpha=0.2
)

plt.title("temperature vs hourly bike rentals")
plt.xlabel("normalized temperature")
plt.ylabel("hourly rentals")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
print("Temperature-demand correlation:", round(df["temp"].corr(df["cnt"])))
```
